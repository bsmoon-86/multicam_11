#  머신러닝 
- 종류
    - 지도 학습
        - 선형 모델
            - 데이터는 모두 선으로 예측이 가능하다. (추론에서부터 시작)
            - 입력 특성(독립 변수)에 따라 선형 함수를 만들어서 예측하는 알고리즘 
        - 트리 모델
            - 조건식을 이용하여 데이터를 분할하고 불순도가 낮을때까지 계속적으로 데이터를 분할하여 예측하는 방법
            - 불순도 : 지니계수, 엔트로피, log_loss
    - 비지도 학습 : 정답이 없는 데이터셋을 이용 
        - 차원 축소 
            - PCA, t-SNE
            - 피쳐가 개수가 너무 많은 경우 데이터의 계산에서 시간이 너무 소모가 되기 때문에 시간을 줄이기 위해서 차원을 축소 
            - 다차원의 데이터를 그래프 시각화 하기에 어려움이 존재하기 때문에 차원 축소를 이용하여 그래프 시각화
            - 텍스트데이터를 토큰화을 한 뒤 단어 간의 유사성을 보여주기 위해서도 사용
        - 군집화 
            - 비슷한 유형의 데이터들을 모아서 하나의 그룹으로 생성 
            - 군집의 개수를 지정할 수 있는 K-Means
                - 중심점을 랜덤하게 생성하고 주변의 데이터들의 위치로 이동하면서 군집을 생성 
            - 군집의 개수를 모델이 정해주는 DBSCAN
                - 밀도 기반의 군집화 
                - 고밀도 지역의 데이터들을 하나의 군집으로 생성 
                - 경계의 데이터들이 군집을 형성할수 있는가? 확인 
                - 형성이 가능하면 군집의 크기를 늘려준다. 
                - 군집에 포함되어있지 않은 데이터는 노이즈로 판단 

### 지도 학습 모델의 종류 
1. 선형 회귀 모델 (LinearRegression)
    - 선형 모델의 기본 모델
    - 독립 변수들의 피쳐들 간의 선형 함수를 생성하여 결과를 예측하는 모델 
    - 비선형 데이터를 예측하려면 성능이 좋지 않다. 
    - 비선형 데이터를 예측하기 위해서 다항 회귀 (PolynomialFeatures)
        - 피쳐의 개수를 늘려준다. (상호관계에 피쳐, 제곱항, 절편항)
2. 가중치에 패널티를 줌으로써 예측도를 올려 주는 모델 
    - 릿지(Ridge)
        - L2 패널티를 이용하여 가중치에 변화를 줌으로써 모델의 성능을 올려준다. 
        - 패널티의 강도가 클수록 회귀 계수가 작아지고 과적합을 방지, 과소적합의 위험 증가 
    - 라쏘(Lasso)
        - L1패널티를 이용하여 가중치에 변화를 주는 모델 
        - 가중치의 변화가 규제 강도에 따라 0까지 감소 
            - 컬럼의 개수가 매우 많은 경우 컬럼을 선택할수 있도록 도와준다. (feature selection)
            - 모델이 단순화의 가능성 존재 
    - 엘라스틱넷(ElasticNet)
        - 릿지, 라쏘 를 절충한 모델 
        - 두개의 패널티를 혼합하여 모델의 성능을 올린다. 
        - l1_ratio 매개변수를 이용하여 두 패널티의 비율을 설정
3. 로지스틱회귀(LogsticRegression)
    - 이름은 회귀지만 결과는 분류 모델 
    - 0과 1사이의 값(실수)을 예측하여 0.5보다 큰 경우라면 1 아니라면 0을 출력하는 알고리즘 
    - 종속 변수가 범주형인 경우 적용하는 회귀 분석 방식
    - 종속 변수 y를 직접 모델링에 활용하지 않고 y가 특정 범주에 속하는 확률을 모델링
    - 분류형 모델들은 데이터의 불균형에 대한 매개변수 class_weight 존재 
4. 서포트벡터머신 (SVM)
    - 데이터를 분리하는 최적의 경계(결정 경계)을 찾고 해당 경계선에서 어느 정도의 마진(여유공간) 생성
    - 회귀의 경우에는 마진 밖에 데이터들을 공간 안으로 들어올수 있도록 패널티 부여 
    - 분류의 경우에는 마진 영역 안쪽의 데이터들을 공간 밖으로 빠질수 있도록 패널티 부여 
5. 의사결정나무(DecisionTree)
    - 나무 구조를 도표화하여 분류와 예측을 수행하는 분석 방법 
    - 적절한 규칙(조건식)을 찾아서 데이터 분할하여 나무를 성장하는 과정 
    - 불순도를 측정하여 불순도가 낮을 방향으로 자식 가지를 형성 
    - 나무의 깊이가 깊어질수록 불순도를 내려가지만 과적합의 위험성이 존재 
- 앙상블 
    - 단일 결정 트리의 단점을 극복하고자 여러 ML 모델을 연결하여 더욱 강력한 모델을 생성 
    - 대표적인 방법 : 배깅, 부스팅
6. 배깅
    - 대표적인 모델 : 랜덤포레스트
    - 부트스트랩 자료를 자료를 생성 ( 전체 데이터에서 일부의 데이터를 가져와서 모델에 입력 )
    - 여러 모델들을 각각 독립적으로 학습으로 하고 결과를 다수결의 원칙에 따라서 다수의 의견을 최종 결과물로 만들어준다. 
    - 부트스트랩 자료, 컬럼의 선택을 하는 이유는? -> 각 모델들에 데이터의 다양성을 가질수 있도록 하기 위함
    - 부트스트랩 자료를 통해 선택 받지 못한 데이터들을 OOB(Out Of Bag)를 이용하여 검증에서 사용이 가능
    - 장점 : 모델의 불안정성을 줄이고 분산이 감소
    - 단점 : 편향은 그대로 유지 
7. 부스팅
    - 대표적인 모델 : XGBoost, LightGBM, CatBoost
    - 여러개의 모델들을 직렬화하여 하나의 모델의 결과를 이용하여 오차를 계산하고 오차를 줄이기 위해 다음 모델을 사용하는 형태
    - 배깅 모델에 비해서 직렬화로 구성되어있기 때문에 시간이 오래 걸린다. 
    - 장점 : 편향 감소
    - 단점 : 순차적인 학습을 통해 학습 시간이 증가 , 학습의 횟수가 너무 많은 경우 과적합, 매개변수 튜닝이 매우 중요

### 교차 검증 
1. KFold
    - 데이터를 k개의 동일 크기로 나눠준다(폴드화)
    - k번 반복하여 하나의 폴드는 검증용으로 나머지 폴드는 학습용으로 사용
    - 모든 폴드는 한번씩 검증으로 사용 
    - 데이터의 수가 적은 경우에는 모델의 성능을 안정적으로 평가 할수 있는 방법 
    - 계층화 Kflod도 존재 
2. GridSearchCV
    - 매개변수의 조합을 생성하고 최적의 조합의 값을 찾는 방법 
    - 조합 별로 모델을 학습하고 검증하여 평가 지표를 통해서 좋은 모델의 매개변수의 조합을 확인
3. Pipeline
    - 머신러닝의 학습 과정을 하나의 연속된 흐름으로 묶는 도구 
    - 데이터의 전처리 과정, 모델의 학습, 예측 과 같은 흐름을 하나로 연결하는 기능 
    - 장점 : 코드 간결화 및 자동화, 데이터의 누수 방지 , 재사용성이 높음 
    - 단점 : 오류 추적이 문제, 학습 시간이 증가, 순차적인 진행이기때문에 병렬화의 어려움

### 평가 지표
- 회귀 모델 
    - MAE
        - 실젯값과 예측값의 차이를 절대값으로 변환하여 평균을 구한 값
    - MSE
        - 실젯값과 예측값의 차이를 제곱하여 평균을 구한 값
        - 실젯값과 예측값의 면적을 의미 
    - RMSE
        - MSE값에 루트를 씌운 값
        - 오차를 제곱하면 오차가 클수록 기하급수적으로 증가하는 상황이 발생하기 때문에 해당 현상을 상쇄하기 위해 사용
    - MSLE
        - MSE값에 로그를 씌운 값
        - RMSE와 같이 기하수급적으로 커지면 상황은 방지하기 위해 사용
    - MAPE
        - MAE를 퍼센트로 표시 
    - R2-Score
        - 1 - (SSE / SST)
        - SSE : 설명하지 못하고 남은 에러 
        - SST : 전체 분산
        - 1에 가까울수록 모델의 성능이 좋다. 
- 분류 모델
    - 혼동행렬
        - 분류 모델에서 예측 오류가 얼마인가 어떠한 유형의 오류가 발생하는가를 나타내는 지표 
        - TN : 예측이 부정, 실제가 부정
        - FP : 예측이 긍정, 실제가 부정 
        - FN : 예측이 부정, 실제가 긍정 
        - TP : 예측이 긍정, 실제가 긍정 
    - 정확도 
        - 실제의 데이터에서 예측 데이터가 얼마나 같은가 판단 
        - TN + TP / TN + FN + FP + TP
    - 정밀도 
        - 긍정으로 예측한 것들 중 실제가 긍정인 비율
        - TP / TP + FP
    - 재현율
        - 실제 긍정인 것들 중 예측이 긍정인 경우 
        - TP / TP + FN
    - F1-Score
        - 정밀도와 재현율을 결합한 분류 성능 지표
        - 2 * TP / 2 * TP + FP + FN
