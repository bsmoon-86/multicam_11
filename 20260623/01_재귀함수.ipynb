# 재귀 함수 
- 함수가 자기 자신을 다시 호출하는 방법 
    - 기동 조건 : 언제 멈출 것인가를 반드시 지정 
    - 재귀 단계 : 문제를 똑같은 형태의 더 작은 문제로 쪼개어 자기 자신을 호출하는 부분 

```python
# 카운트다운 ->  3 2 1 발사 
for i in range( 3, -1, -1 ):
    if i != 0:
        print(i)
    else:
        print('발사')
```

```python
def countdown(n):
    # n : 카운트의 시작 수치 
    # 기동 조건  : n이 0이 되면 함수 호출을 그만 둔다. (발사 출력을 생성)
    if n == 0:
        print('발사')
        return
    # 수행하는 작업 
    print(n)

    # 재귀 단계 : n에서 1을 뺀 값을 이용하여 자기 자신을 다시 호출한다. 
    countdown(n - 1)

countdown(3)
```

```python
def sum_list(arr):
    # arr에 대입이 되는 원본의 데이터를 유지 
    # 재귀가 멈추는 조건 -> arr의 길이가 0이 되는 경우 
    if len(arr) == 0:
        return 0
    
    return arr[0] + sum_list(arr[1:])

sum_list(_list)
```

```python
# 피보나치 수열 : 1, 1, 2, 3, 5, 8, 13, 21, ....

def fibonacci(n):
    # n : 피보나치 수열의 몇번째 데이터를 확인할것인가?
    if n == 1 or n == 2:
        return 1
    return fibonacci(n-1) + fibonacci(n-2)

fibonacci(5)
```

```python
# 리버스 -> 뒤에서부터 하나씩 뽑는다. (모든 값은 다 뽑았을때 -> 길이가 0이 됬을때)
def reverse_custom(data):
    if len(data) == 0:
        return data
    
    return data[-1] + reverse_custom(data[:-1])

text = 'python'
reverse_custom(text)
```

```python
list_2 = [1, [2, 3], [4, [5, 6]], 7]
# list_2를 1차원으로 평탄화 작업 
result3 = []
result2 = []
result1 = []
for item in list_2:
    # print(item)
    # item의 타입이 list면 반복문을 한번 더 사용
    if type(item) == list:
        for item2 in item:
            if type(item2) == list:
                for item3 in item2:
                    result3.append(item3)
            else:
                # print(item2)
                result2.append(item2)
    else:
        # print(item)
        result1.append(item)
    print(result1, result2, result3)
```

```python
def flat(data):
    # data : 리스트 형태의 데이터
    result = []

    for item in data:
        if type(item) == list:
            # flat() -> 결과 리스트 생성 
            # result.extend(flat(item))
            result += flat(item)
        else:
            result.append(item)

    return result

flat(list_2)
```

```python
import random
```

```python
# 로또 점수 자동 생성 -> 1 - 45 까지의 숫자를 중복 없이 숫자를 생성하는 재귀 함수 
lotto_list = list(range(1, 46, 1))

# 숫자 6개를 중복 없이 출력하는 재귀함수 
def lotto(num_list = []):
    global lotto_list
    result = num_list
    # 결과는 6개의 숫자 함수를 종료
    if len(num_list) == 6:
        # print(num_list)
        return result

    # lotto_list에서 무작위한 값을 하나 선택 
    random_int = random.choice(lotto_list)

    if random_int not in num_list:
        num_list.append(random_int)
    # print(num_list)
    return lotto(num_list)

```

```python
api_response = {
    "store_name": "맛있는 치킨 본점",
    "categories": [
        {
            "category_name": "추천 메뉴",
            "items": [
                {"item_name": "황금 올리브 치킨", "price": 20000},
                {"item_name": "양념 치킨", "price": 21000}
            ]
        },
        {
            "category_name": "사이드 메뉴",
            "items": [
                {
                    "item_name": "치즈볼 세트", 
                    "price": 5000,
                    "options": [
                        {"option_name": "크림치즈 변경", "price": 5500},
                        {"option_name": "모짜렐라 곱빼기", "price": 6000}
                    ]
                },
                {"item_name": "감자튀김", "price": 4000}
            ]
        }
    ]
}
```

```python
def flat_menu_data(data, menu_name = '일반 메뉴'):
    result = []
    # data의 타입이 dict 인 경우
    if type(data) == dict:
        # item_name이라는 키 값이 존재하는가?
        if 'item_name' in data:
            menu_name = data['item_name']
        
        # item_name이 존재하거나 option_name이 있다면 행을 추가한다. 
        if 'item_name' in data:
            result.append(
                {
                    '이름' : menu_name, 
                    '가격' : data['price']
                }
            )
        elif 'option_name' in data:
            result.append(
                {
                    '이름' : f"{menu_name} ( {data['option_name']} )", 
                    '가격' : data['price']
                }
            )
        
        # 하위 딕셔너리가 존재하는 경우 내부의 데이터를 재귀함수를 이용하여 재 호출 
        for key, value in data.items():
            # value들이 list 이거나 dict 인 경우
            if isinstance( value, (list, dict) ):
                # 함수 재 호출 
                result.extend(flat_menu_data( value, menu_name ))
    
    # data의 타입이 list 인경우 (각각의 원소들을 재귀 호출을 이용하여 대입 )
    elif type(data) == list:
        for item in data:
            result.extend(flat_menu_data( item, menu_name ))
    
    return result
```

```python
# 알고리즘 문제 
# 체스판(8*8)에 퀸 8개를 배치하여 모든 퀸이 다른 퀸들을 공격 할수 없는 경우의 수는 몇개인가? 

# 체스판에 퀸들의 배치할수 있는 모든 경우 
num = 3

pos = [0] * num

def put():
    # 각 열에 배리한 퀸의 위치를 출력 
    for j in range(num):
        for i in range(num):
            print('o' if pos[i] == j else 'x', end = ' ')
        # 출력의 줄바꿈
        print()
    # 하나의 배치가 종료했으면 줄바꿈 1회 추가
    print()

def set(i):
    # i 열에 퀸을 배치
    for j in range(num):
        pos[i] = j
        if i == num - 1:
            put()
        else:
            set(i + 1)

set(0)
```

```python
num = 4

pos = [0] * num
# 경우의 수 초기 값
cnt = 0

# 퀸의 자리가 안전한가? 
def is_safe(row, col):
    for i in range(row):
        # 같은 열에 퀸이 존재하는가?
        if pos[i] == col:
            return False
        
        # 대각선에 다른 퀸이 존재하는가? (행의 차이와 열의 차이가 같은 경우가 대각선 )
        if abs(pos[i] - col) == abs(i - row):
            return False
        
    return True

# row번째 행에 퀸을 배치하는 재귀 함수 
def solve_queens(row):
    global cnt

    # 0번부터 num-1번 행까지 모두 성공적으로 배치하여 n번째 행에 도달하는 경우 
    if row == num:
        cnt += 1
        print(f"{cnt}번째 정답 모드 상태")
        print(pos)
        return

    for col in range(num):
        if is_safe(row, col):
            pos[row] = col

            # 다음 행으로 들어간다. 
            solve_queens(row+1)

            # 다음 행으로 이동을 했는데 실패해서 return 구문으로 돌아온다면 다음 col반복문으로 넘어가면서 이전 상태로 되돌아간다. 
solve_queens(0)
print(cnt)
```
