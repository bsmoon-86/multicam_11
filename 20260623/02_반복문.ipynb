### 반복문 
- 소수 나열하기 
- 소수?
    - 자기 자신과 1 이외에는 나누어 떨어지지 않은 수를 의미 
        - 조건식 : 2부터 시작해서 n-1 까지 어떤 정수로도 나누어 떨어지지 않는다(나머지가 0인 정수가 존재하지 않는다.)

```python
# 2부터 1000까지의 정수 중 소수의 리스트를 출력하시오
# 나눗셈의 횟수도 출력

cnt = 0

prime_list = []

# 2부터 1000까지의 반복을 하는 반복문을 생성 
for n in range(2, 1001):

    # 조건식 : 1과 자신 자신을 제외한 값으로 나누었을때 딱 떨어지지 않는다. 
    for i in range(2, n):
        cnt += 1
        if n % i  == 0:
            break
    else:
        prime_list.append(n)

print("나눗셈의 횟수 : ", cnt)
print("소수의 배열 : ", prime_list)
print("소수의 갯수 : ", len(prime_list))
```

```python
# 짝수는 소수가 아니다. (2는 제외)

cnt = 0
prime_list = [2]

for n in range(3, 1001, 2):
    for i in range(2, n):
        cnt += 1
        if n % i == 0:
            break
    else:
        prime_list.append(n)

print(cnt)
print(len(prime_list))
```

```python
# 소수의 조건식 : 이미 찾은 소수로 나눠서 나머지가 0이라면 소수가 아니다. 

cnt = 0
prime_list = [2]

for n in range(3, 1001, 2):
    for i in range(1, len(prime_list)):
        cnt += 1
        if n % prime_list[i] == 0:
            break
    else:
        prime_list.append(n)

print(cnt)
print(len(prime_list))
```

```python
# n의 제곱근 이하의 어떤 소수로도 나누어 떨어지지 않는다. 
# 조건식 -> prime_list[i] ** 2 <= n 인 경우에은 소수 데이터 n을 나눠서 확인 

cnt = 0
prime_list = [2, 3]

for n in range(5, 1001, 2):
    i = 1
    while prime_list[i] ** 2 <= n:
        cnt += 2
        if n % prime_list[i] == 0:
            break
        # while문은 조건식이 거짓이 될수 있도록 특정 값을 조정
        i += 1
    else:
        prime_list.append(n)
        cnt += 1

print(cnt)
print(prime_list)
print(len(prime_list))
```
