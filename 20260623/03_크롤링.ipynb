### 크롤링 
- 웹페이지(html문서)에서 특정 영역의 텍스트를 긁어서 수집하는 과정 
    - html문서 
        - 구조를 나타내는 html언어 (크롤링에서 봐야되는 부분)
        - 디자인을 나타내는 css언어
        - 페이지에서 동적 작동을 시켜주는 js 언어 
        - `<head>`
            - 외부의 파일을 참조
            - meta data
            - 탭의 이름을 지정 
        - `<body>`
            - 실제 화면에서 나타내는 부분 
            - 크롤링에서 중요하게 봐야되는 부분 
        - `<footer>`
            - 화면의 마지막 부분을 담당 
            - 일반적으로는 회사의 정보나 페이지에 대한 정보를 담당하는 부분
- 크롤링에서 사용하는 라이브러리 
    - requests
        - 특정 주소에 요청을 보내고 응답을 받는 라이브러리 
        - get() 함수를 이용하여 요청을 보내고 응답을 받는다. 
            - params 매개변수 
                - 특정 주소에 데이터를 같이 실어서 보낸다. 
                - 데이터의 형태는 dict 구조로 보낸다. 
            - header 매개변수
                - 데이터를 보내는 방식 
                - head 공간에 데이터를 실어서 보낸다. 
        - 응답 데이터의 형태
            - json, xml 
                - 오픈 data api
            - html문서
                - 웹 서비스 페이지
    - bs4
        - BeautifulSoup() 클래스를 이용
            - 문자화 되어있는 HTML문서를 html.parser를 이용하여 type을 변경(파싱)
            - 타입을 변경(class 안에 새로운 데이터를 담아서 객체를 생성)
            - HTML 내의 태그들에 접근을 편하게 사용하기 위함
            - 내장 함수 
                - 찾는 태그중 첫번째 태그를 찾는 함수 
                    - find()
                    - soup.태그명
                    - 함수의 결과 값의 타입은 Tag
                - 찾는 태그 중 조건에 맞는 모든 태그를 찾는 함수 
                    - find_all()
                    - 함수의 결과 값의 타입은 ResultSet
                        - ResultSet은 Tag들이 모여있는 리스트
                - 태그 안에서 content(실제 화면에 보이는 텍스트)를 추출하는 함수 
                    - get_text()
                        - 현재 태그 외의 자식 태그들의 텍스트도 같이 추출
                    - 태그.string
                        - 현재 태그의 텍스트만 추출 
    - selenium 
        - 웹 브라우져를 code를 이용해서 제어하는 기능 
        - 크롤링에서 selenium을 사용하는 이유는?
            - 동적인 웹 페이지의 내용을 가져오는 위함(비동기 방식)
        - webdriver
            - 브라우져를 불러오는 기능 (Chrome, Edge, Safari, ...)
        - find_element()
            - 브라우져의 html문서에서 조건에 맞는 태그 첫번째를 선택 
            - BeautifulSoup에서 find()와 흡사
        - find_elements()
            - 브라우져의 html문서에서 조건에 맞는 태그 모두를 선택 -> []
            - find_all()과 흡사 
        - By
            - 태그의 조건을 생성하기 위해 불러오는 기능 
            - ex) By.ID, id명 --> 태그중 ID값이 id명인 태그
        - Keys
            - 키보드 특수 이벤트를 불러오는 기능 
            - ex) Keys.ENTER -> 키보드의 enter 이벤트 발생
        - send_keys()
            - 특정 태그에 이벤트를 발생 
            - 문자를 입력, 키보드 특수 이벤트를 발생 
        - click()
            - 선택된 태그에 마우스 이벤트 좌클릭 발생
        - page_source
            - 브라우져의 탭의 html문서를 python 환경에 문자열로 되돌려받는 기능
        - 클릭 이벤트 후 탭이 여러개 생성이 되었을때 
            - driver.window_handles
                - 오픈된 브라우져의 탭들의 주소값을 확인 
                - 탭을 이동하기 전 탭들의 주소를 확인
            - driver.switch_to.window()
                - 이동하려는 주소 값을 인자로 넣어서 호출하면 탭이 이동 
                - driver.switch_to.window(
                    driver.window_handle[1]
                )
        - execute_script()
            - JS로 이루어진 code를 실행
            - 마우스 휠 이벤트 발생 ( 'window.scrollBy(0, 800);' )
- pandas를 이용한 크롤링
    - read_html()
