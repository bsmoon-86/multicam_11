#### Okt
- SNS, 댓글등의 한글로 이루어진 데이터를 형태소 분석하기 위한 분석기 
- 주요 품사 태그 종류
    - Noun : 명사 (데이터, 분석)
    - Verb : 동사 (하다, 먹다 )
    - Adjective : 형용사 (재미있다, 좋다)
    - Josa : 조사 (은, 는, 이, 가)
    - Adverb : 부사 (정말, 매우)
    - KoreanParticle : 한국어 파티클 / 이모티콘 (예 : ㅋㅋㅋ, ㅠㅠ)
    - Punctuation : 구두점 (.   , ? ~)
    - Number : 숫자
    - Alpha : 영문자
    - Modifier : 관형사 (새, 헌, 첫)
    - Foreign : 외국어 / 한자
    - ScreenName : 트워터 아이디 표현(@username)
    - Hashtag : 해시태그(#태그)
- 매개변수
    - 정규화 (norm)
        - 문자를 정규화
        - 비표준어, 오타, 불필요한 반복 등을 깔끔하게 다듬어준다. 
        - 문자들에서 노이즈를 제거 
    - 어간 추출 (stem)
        - 사전에 등록된 기본 형태만 남기는 방법
        - 한국어의 동사와 형용사는 문맥에 따라서 형태가 매우 많다. 꼬리(어미)를 잘라내고, 기본형으로 통일 

```python
from konlpy.tag import Okt
```

```python
okt = Okt(jvmpath=r'C:\Program Files\Java\jdk-17')
```

```python
text = "파이썬에서 자바의 환경 변수 설정은 너무 힘이 듭니다. ㅠㅠ 오프라인 수업이 그립습니닼ㅋㅋ"

# 형태소 분리 
tokens = okt.morphs(text)
print(tokens)
```

```python
text.split()
```

```python
text2 = "파이썬에서자바의환경변수설정은너무힘이듭니다ㅠㅠ오프라인수업이그립습니다"

tokens2 = okt.morphs(text2)
print(tokens2)
```

```python
text2.split()
```

```python
# 명사 추출 
nouns = okt.nouns(text)
print(nouns)
```

```python
# 품사 태깅 
# 단어와 해당 단어의 품사를 튜플의 형태로 묶어서 반환 
print(okt.pos(text))
```

```python
# Josa, Punctuation 품사를 제외하고 나머지 문자들만 사용
new_list = []

for t, p in okt.pos(text):
    # t : 텍스트, p : 품사명
    # print(t)
    # print(p)
    # break
    if p not in ['Josa', 'Punctuation']:
        new_list.append(t)
new_list
```

```python
# 정규화 , 어간 추출 
# norm : 문자의 오타, 비표준어, 불필요한 반복을 다듬어주는 부분
# stem : 꼬리(어미)를 잘라내고 기본형으로 변환
okt.pos(text, norm=True, stem = True)
```

### kkma 분석기 
- 굉장히 깐깐한 언어 분석기 
- 국어사전에 나온 단어들을 이용하여 단어를 토큰화 하는 분석기

#### 세종 태그셋
- 명사류
    - NNG : 일반 명사 (데이터, 분석)
    - NNP : 고유 명사 (서울, 컴퓨터)
    - NNB : 의존 명사 (것, 수, 만큼)
    - NP : 대명사 (나, 너, 우리)
- 동사 / 형용사
    - VV : 동사 (배우다, 먹다)
    - VA : 형용사 (좋다, 빠르다)
- 조사
    - JKS : 주격 조사 (이, 가)
    - JKO : 목적격 조사 (을, 를)
    - JKM : 부사격 조사 (에, 에서, 으로)
- 기타
    - MAG : 일반 부사 (매우, 잘, 빨리)
    - EFM : 평서형 종결 어미 (~다, ~는다)
    - SF : 마침표, 물음표, 느낌표


```python
# 꼬꼬마
from konlpy.tag import Kkma
```

```python
kkma = Kkma()
```

```python
# 문장 분리용 샘플 텍스트 생성 
test_formal = "인공지능 데이터 분석가 과정에서 오신것을 환영합니다 오늘부터 자연어 처리를 배우고 다음 주에는 딥러닝 자연어 처리를 배울 예정입니다 끝까지 포기하지 마세요"

# 문장을 분리 (sentences)
# 구두점이 존재하지 않더라도 문맥을 파악해서 문장을 리스트로 생성 
sentences = kkma.sentences(test_formal)
print(sentences)
```

```python
# 형태소 분리 
text = "파이썬에서 자바의 환경 변수 설정은 너무 힘이 듭니다. ㅠㅠ 오프라인 수업이 그립습니닼ㅋㅋ"
kkma.morphs(text)
```

```python
# 명사 추출 
print(kkma.nouns(text))
```

```python
# 품사 태깅
print(kkma.pos(text))
```
